````python
import numpy as np

from sklearn.datasets import make_blobs, make_moons
from sklearn.model_selection import train_test_split

'''
    Try changing the number of samples, the cluster centers, and how spread out each cluster is.
    You’ll notice that k-means does not perform well when clusters have different sizes or densities.
    The inital values are just `dummy` values
'''
samples = np.array([1, 1, 1, 1, 1])
cluster_centers = np.array([
    [-1,1],
    [0,5],
    [2, -2.5],
    [-2, -1],
    [0, -2],
])
# Try Changing Up the Standard Deviations for Each Cluster
cluster_std = np.array([0.5, 0.5, 0.5, 0.5, 0.5])
X, y, centers = make_blobs(n_samples=samples, centers=cluster_centers, cluster_std=cluster_std, shuffle=True, random_state=64, return_centers=True)

y[samples[0]] = 1
y[samples[1]] = 2
y[samples[2]] = 3
y[samples[3]] = 4
y[samples[4]] = 5
````

In k-means, “labels” just mean cluster numbers, not real answers because there is no ground-truth in k-means


````python
'''
    This snippet of code for plotting the clusters with their
    decision boundaries and centroids is accredited to
    Aurélien Géron from, "Hands-On Machine Learning with
    SciKit-Learn, Keras, and TensorFlow, 3rd Edition
    ISBN-13: 9781098125974
    E-text ISBN-13: 9781098122461
'''
def plot_data(X):
    plt.plot(X[:, 0], X[:, 1], 'k.', markersize=2)

def plot_centroids(centroids, weights=None, circle_color='w', cross_color='k'):
    if weights is not None:
        centroids = centroids[weights > weights.max() / 10]
    plt.scatter(centroids[:, 0], centroids[:, 1],
                marker='o', s=35, linewidths=8,
                color=circle_color, zorder=10, alpha=0.9)
    plt.scatter(centroids[:, 0], centroids[:, 1],
                marker='x', s=2, linewidths=12,
                color=cross_color, zorder=11, alpha=1)

def plot_decision_boundaries(clusterer, X, resolution=1000, show_centroids=True,
                             show_xlabels=True, show_ylabels=True):
    mins = X.min(axis=0) - 0.1
    maxs = X.max(axis=0) + 0.1
    xx, yy = np.meshgrid(np.linspace(mins[0], maxs[0], resolution),
                         np.linspace(mins[1], maxs[1], resolution))
    #Fill the whole space with tiny points--> ask the model what cluster each point belongs to --> color the space accordingly
    Z = clusterer.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)

    plt.contourf(Z, extent=(mins[0], maxs[0], mins[1], maxs[1]),
                cmap="Pastel2")
    plt.contour(Z, extent=(mins[0], maxs[0], mins[1], maxs[1]),
                linewidths=1, colors='k')
    plot_data(X)
    if show_centroids:
        plot_centroids(clusterer.cluster_centers_)

    if show_xlabels:
        plt.xlabel("$x_1$")
    else:
        plt.tick_params(labelbottom=False)
    if show_ylabels:
        plt.ylabel("$x_2$", rotation=0)
    else:
        plt.tick_params(labelleft=False)

plt.figure(figsize=(8, 4))
plot_decision_boundaries(kmeans, X)
plt.show()
````

## DBSCAN

[DBSCAN](https://scikit-learn.org/stable/modules/generated/sklearn.cluster.DBSCAN.html) by SciKit-Learn

````python
'''
    Adjust the noise, amount of samples from make_moons; additionally
    with DBSCAN adjust the values for epsilon (eps) and min_samples
    to see how it changes the results.
'''
from sklearn.cluster import DBSCAN
# Code to be completed, change the values where it says, "None"
X, y = make_moons(n_samples=None, noise=None, random_state=64)
dbscan = DBSCAN(eps=None, min_samples=None)
dbscan.fit(X)
````

````python
for x in dbscan.labels_[:25]:
    print(x, end= " ") #The cluster assignment for each data point, -1 means that the point is considered an outlier
````

````python
for x in dbscan.core_sample_indices_[:25]:
    print(x+1, end= " ") #The indices (positions) of "core" points
````

````python
for x, y in dbscan.components_[:25]:
    print(f"({x.round(4)}, {y.round(4)})", end=" ") #The actual coordinates of the core points
````

````python
'''
    This snippet of code for plotting the clusters with their
    decision boundaries and centroids is accredited to
    Aurélien Géron from, "Hands-On Machine Learning with
    SciKit-Learn, Keras, and TensorFlow, 3rd Edition
    ISBN-13: 9781098125974
    E-text ISBN-13: 9781098122461
'''
def plot_dbscan(dbscan, X, size, show_xlabels=True, show_ylabels=True):
    core_mask = np.zeros_like(dbscan.labels_, dtype=bool)
    core_mask[dbscan.core_sample_indices_] = True
    anomalies_mask = dbscan.labels_ == -1
    non_core_mask = ~(core_mask | anomalies_mask)

    cores = dbscan.components_
    anomalies = X[anomalies_mask]
    non_cores = X[non_core_mask]

    plt.scatter(cores[:, 0], cores[:, 1],
                c=dbscan.labels_[core_mask], marker='o', s=size, cmap="Paired")
    plt.scatter(cores[:, 0], cores[:, 1], marker='*', s=20,
                c=dbscan.labels_[core_mask])
    plt.scatter(anomalies[:, 0], anomalies[:, 1],
                c="r", marker="x", s=100)
    plt.scatter(non_cores[:, 0], non_cores[:, 1],
                c=dbscan.labels_[non_core_mask], marker=".")
    if show_xlabels:
        plt.xlabel("$x_1$")
    else:
        plt.tick_params(labelbottom=False)
    if show_ylabels:
        plt.ylabel("$x_2$", rotation=0)
    else:
        plt.tick_params(labelleft=False)
    plt.title(f"eps={dbscan.eps:.2f}, min_samples={dbscan.min_samples}")
    plt.grid()
    plt.gca().set_axisbelow(True)

dbscan2 = DBSCAN(eps=0.2)
dbscan2.fit(X)

plt.figure(figsize=(9, 3.2))
plt.subplot(121)
plot_dbscan(dbscan, X, size=100)

plt.subplot(122)
plot_dbscan(dbscan2, X, size=100, show_ylabels=False)
plt.show()
# Left plot → 2 clusters (but messy, lots of noise, red x points)
# Right plot → single cluster
````

### Performing Classification with DBSCAN + KNN

[KNN](https://scikit-learn.org/stable/modules/generated/sklearn.neighbors.KNeighborsClassifier.html) by SciKit-Learn

### **The DBSCAN-to-KNN/SVM Prediction Pipeline**

```text
=================== PHASE 1: UNSUPERVISED CLUSTERING ===================

  +-------------------+
  | 1. Training Data  | (Unlabeled X_train)
  +-------------------+
            |
            v
  +-------------------+
  |    2. DBSCAN      | (Discovers dense regions)
  +-------------------+
            |
            v
  +-------------------+
  | 3. Labeled Data   | (Points assigned to clusters: 0, 1, 2... & Noise: -1)
  +-------------------+

=================== PHASE 2: SUPERVISED TRAINING =======================

            |
            v
  +-------------------+
  | 4. Preprocessing  | (Crucial Step: Handle Noise points)
  +-------------------+
            |
            v
  +-------------------+
  |  5. Train KNN/SVM | (Features = X_train, Target = DBSCAN Labels)
  +-------------------+
            |
            v
  +-------------------+
  |  6.Trained KNN/SVM| (Model with established decision boundaries)
  +-------------------+

=================== PHASE 3: NEW DATA PREDICTION =======================

  +-------------------+        +-------------------+        +-------------------+
  |  New Data Point   | -----> |   Trained KNN/SVM | -----> | Predicted Cluster |
  |   (Unseen Data)   |        |   (Classifier)    |        |      Label        |
  +-------------------+        +-------------------+        +-------------------+

````python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=100)
knn.fit(dbscan.components_, dbscan.labels_[dbscan.core_sample_indices_])
````

````python
X_new = np.array([
    [0,0],
    [1,-0.25],
    [1.2,-0.3],
    [0.4,0.5],
    [-0.867, -0.6]
                 ])
````

````python
for y_pred in knn.predict(X):
    print(y_pred, end="  ") # this is the prediction for the data points in X, which is the original dataset.
````

````python
knn.predict_proba(X)
````

````python
for y_pred_new in knn.predict(X_new):
    print(y_pred_new, end="  ") # this is the prediction for the new data points in X_new, which is the new dataset.
````
